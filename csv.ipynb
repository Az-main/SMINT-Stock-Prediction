```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('Samsung_stock.csv')
df['Date'] = pd.to_datetime(df['Date'])
# print(df['Date'].dtype)
start= '2019/10/2'
end = '2020/1/1'

df_filtered = df[(df['Date'] >= start) & (df['Date'] <= end)]

# Ensure the 'open' column is numeric for plotting
df_filtered['open'] = pd.to_numeric(df_filtered['open'], errors='coerce')

sns.lineplot(x=df_filtered['Date'], y=df_filtered['open'])
plt.show()
```
